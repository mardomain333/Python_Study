🔹 4️⃣ Array Creation Methods

Creating arrays is extremely important because everything in NumPy starts from this.

There are 5 main ways to create arrays:

From Python objects

Using predefined shape functions

Using range-based functions

Using random functions

From existing data (copying / reshaping / reading buffers)

Let’s go one by one.

```python
#4.1 Creating Arrays from Python Objects

#we can convert our python list into numpy array

arr=np.array([[1,2,3],[1,2,0]])
print(arr.ndim,arr.size,arr.shape)
# What happens internally?

# Python list → converted to contiguous C array

# All elements converted to same dtype

# Stored in contiguous memory block 
arr=np.array([1,2.5,3,5])
print(arr) #every thing converted into float
```

```python
# #4.2 Predefined Shape Functions

# These functions create arrays without manually writing values.
np.zeros((2,3))
np.ones((2,2))
np.full((2,2), 7)
np.eye(3)
np.identity(3)

arr=np.zeros((3,3))
print(arr)

arr=np.ones((2,2))
print(arr) 

arr=np.full((3,4),7)
print(arr)

arr=np.eye(3,k=-2)
print(arr)

arr=np.identity(3)
print(arr)

arr=np.empty((2,6))
print(arr)
#np.empty gives unpredictable values, while np.zeros gives a clean slate of zeros.

```

```python
# 4.3 Range-Based Creation
# 🔹 np.arange()

# Like Python range but returns array.

arr=np.arange(1,10,2)#start,end,step 
print(arr)

arr=np.linspace(0,1,5)  #np.linspace is a NumPy function used to generate evenly spaced numbers over a         specified interval. It’s especially handy when you want a sequence of values between two endpoints.
print(arr)
# In short: use np.logspace when you want values that grow exponentially (common in scientific computations, plotting frequency scales, etc.), while np.linspace is for straight linear spacing.

arr=np.logspace(0,2,5,dtype=int) 
# Generate 5 numbers between 10^0 and 10^2
print(arr)
```

```python
# 4.4 Random Array Creation

import numpy as np
  # reproducible results
np.random.seed(5)
print(np.random.rand(3,2))       # 3 random floats [0,1)
#print(np.random.randint(1, 10, 5))  # 5 random integers
#print(np.random.choice(['A','B','C'], 3))  # random choice
# What is use of random.seed ?
# Computers cannot generate true randomness. Instead they generate: Deterministic numbers that look random
# These numbers are generated using: # A mathematical formula
# Starting from an initial number called the seed
# 🔹 What is a Seed?
# A seed is just a starting number for the random generator.
# Think like this:
# If the seed is 10,
# the generator will always produce the same sequence.
# If the seed is 42,
# it will produce another fixed sequence.

#print(np.random.randn(5,5))
#np.random.randint(1,10,(2,3))

```

```python
# # Indexing and Slicing in NumPy (Very Important)

# 1D Array Indexing
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print(a[1:4:2]) 
b=a[1:4] 

#here it share the view that means if i change b later it will also change in a
print(id(a))
print(id(b))
a=[1,2,3,4,5]
l=a
print(id(a))
print(id(l))

```

```python
# 2D Array Indexing

arr=np.array([[1,2,3],[1,2,3],[1,2,3]])
for i in range(2):
    for j in range(3):
        print(arr[i,j])
print(arr[:,:]) #row,column

print(arr[:,:2]) #limit row to 2 rows limit column to 2
```

```python
#  — Boolean Indexing
arr=np.array([1,2,3,4,5])
print(arr[arr>2])
```

```python
# -Fancy Indexing
print(arr[[0,3]]) #print the elements at the 0 and 3 index
```

```python
# - Reshaping
arr=np.arange(6)
print(arr)
print(arr.shape)
print(arr.reshape(2,3))
```

```python
# Row 1 → 0 1 2
# Row 2 → 3 4 5

# NumPy fills row by row (C-order).
# NumPy can calculate one dimension automatically.

print(arr.reshape(-1, 2))
```

```python
b = np.array([[1,2,3],
              [4,5,6]])
print(b.flatten()) #not a view
print(b.ravel()) #its a view
```
