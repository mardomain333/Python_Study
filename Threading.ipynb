Threading -- doing multiple task at the same time inside one program 
eg dwl a file
   print number 
   wait for user input
--all at once 


```python
#without threading 

import time 
start=time.time()
def task():
    time.sleep(4)
    print("task done..")
task()
task()
end=time.time()
print("time taken",round(end-start,1))
```

```python
#with threading 

import threading
import time

start=time.time()
def task():
    time.sleep(4)
    print("task done..")
t1=threading.Thread(target=task)
t2=threading.Thread(target=task)
t1.start() #to start t1 thread
t2.start() #to start t2 thread
t1.join() #wait until t1 is completed
t2.join() #wait until t2 is completed
end=time.time()
print("time taken",round(end-start,1))
```

```python
#multiple threading using for loop(without threading ...)

def sum_of_square(n):
    sums=0
    for i in range(n):
        sums+=i**2
    return sums

def sleep_for_seconds(seconds):
    time.sleep(seconds)


def main():
    start=time.time()
    for i in range(10):
        maxvalue=i*1000000
        print(sum_of_square(maxvalue))
    print("time taken for squaring=",round(time.time()-start))
    
    start=time.time()
    for i in range(1,6):
        sleep_for_seconds(i)

    print("time taken for sleeping=",round(time.time()-start))





if __name__=='__main__':
    main()
```

```python
#multiple threading using for loop

def sum_of_square(n):
    sums=0
    for i in range(n):
        sums+=i**2
    print(sums)

def sleep_for_seconds(seconds):
    time.sleep(seconds)


def main():
    start=time.time()
    thread_list=[]
    for i in range(1,10):
        maxvalue=i*1000000
        t=threading.Thread(target=sum_of_square,args=(maxvalue,))
        t.start()
        thread_list.append(t)
        # print(sum_of_square(maxvalue))
    for i in range(len(thread_list)):
        thread_list[i].join()
    print("time taken for squaring=",round(time.time()-start))
    start=time.time()
    thread_list=[]
    
    for i in range(1,6):
        t=threading.Thread(target=sleep_for_seconds,args=(i,))
        thread_list.append(t)
        t.start()
        # sleep_for_seconds(i)
    
    for i in range(len(thread_list)):
        thread_list[i].join()
    print("time taken for sleeping=",round(time.time()-start,1))





if __name__=='__main__':
    main()
```
