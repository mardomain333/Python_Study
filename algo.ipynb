3> fixed size sliding window
   find the maximum sum of subarray k

   

```python
arr = [1,2,3,1,2,-3,2,3,3]
k = 3

window_sum = sum(arr[:k])
max_sum = window_sum
l = 0
maxs=arr[l:k]
for r in range(k, len(arr)):
    window_sum += arr[r]
    window_sum -= arr[l]
  
    l += 1
    if window_sum>max_sum:
        maxs=arr[l:r+1]
    max_sum = max(max_sum, window_sum)

print(max_sum)
print(maxs)
```

4> longest substring without repeating characters 
 

```python
s = "abcabcbb"
longest=0
set_repeated=set()
l=0
for r in range(len(s)):
    
    while(s[r] in set_repeated ):
        set_repeated.remove(s[l])
        l+=1
    set_repeated.add(s[r])
    current_len=(r-l)+1
    longest=max(longest,current_len)
print(longest)
```

```python
arr = [5, -2, 3, 4, -1]

max_sum = arr[0]
current_sum = arr[0]

start = end = temp_start = 0

for i in range(1, len(arr)):
    if arr[i] > current_sum + arr[i]:
        current_sum = arr[i]
        temp_start = i
    else:
        current_sum += arr[i]

    if current_sum > max_sum:
        max_sum = current_sum
        start = temp_start
        end = i

print("Max Sum:", max_sum)
print("Subarray:", arr[start:end+1])
print("Indices:", start, end)
```

6> Two sum


```python
def twoSum(nums, target):
    seen = {}

    for i, num in enumerate(nums):
        diff = target - num
        if diff in seen:
            return [seen[diff], i]
        seen[num] = i
nums = [2, 7, 11, 15]
target = 9
a,b=twoSum(nums,target)
print(nums[a:b+1])

#example code 
l=[1,2,2,3,4,5]
target=3
is_there=False
serve={

}
for i in range(len(l)):
    if target-l[i]in serve.keys():
        print(f"index {serve[target-l[i]],i}")
        print(f"values{l[serve[target-l[i]]],l[i]}")
        print("======================================")
        is_there=True
    else:
        serve[l[i]]=i
if not is_there:
    print("no pair ...")


```

# You are given N employees (N is even).
# Each employee can work from Office X or Office Y.

# X[i] = travel cost to Office X

# Y[i] = travel cost to Office Y

# Exactly N/2 employees must go to Office X

# 👉 Find the minimum total travel cost.x

```python

N=2
X = [12, 11]
Y = [10, 7]

total=sum(X)

dif=[Y[i]-X[i] for i in range(N)]

dif.sort()

for i in range(N//2):
    total+=dif[i]

print(total)
```
