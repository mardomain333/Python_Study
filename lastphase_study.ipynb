```python
#fixed size sliding window
#max sum for a substring of lenght k
l=[1,2,3,4,5,-5]
k=4
maxsum=sum(l[:k])
li=0
temp=maxsum
start=0
end=k-1
for r in range(k,len(l)):
    temp-=l[li]
    temp+=l[r]
    li+=1
    if temp>maxsum:
        start=li
        end=r
        maxsum=temp


print(maxsum)
print(l[start:end+1])
```

```python
# You are given:
# n items
# Each item has:
# value v[i]
# weight w[i]
# A knapsack with capacity W
# Find the maximum total value you can carry if you are allowed to take fractions of items.




values = [60, 100, 120]
weights = [10, 20, 30]
W = 50

item=[]

for i in range(len(values)):
    ratio=values[i]/weights[i]
    item.append((ratio,values[i],weights[i]))

item.sort(reverse=True)
total=0
for ratio,value,weight in item:
    if weight<=W:
        total+=value
        W-=weight
    else:
        total+=ratio*W
        break
print(total)
```

```python

def find(x):
    if parent[x]!=x:
       return find(parent[x])
    else:
        return parent[x]

jobs = [
    ('A', 2, 100),
    ('B', 1, 19),
    ('C', 2, 27),
    ('D', 1, 25),
    ('E', 3, 15)
]

jobs.sort(key=lambda x:x[2],reverse=True)
maxdeadline=max(job[1] for job in jobs)
parent=[i for i in range(maxdeadline+1)]
result=[]
profit=0

for job in jobs:
    name,deadline,profits=job
    availabe=find(deadline)
    if availabe>0:
        profit+=profits
        result.append(name)
        parent[availabe]-=1
print(result)
print(profit)
```

```python
from collections import Counter
def shortest_substring(pattern,string):
    pattern_count=Counter(pattern)
    string_count=Counter()
    left=0
    shortest=None
    for right,char in enumerate(string):
        string_count[char]+=1
        while(all(string_count[key]>=pattern_count[key] for key in pattern_count)):
            if shortest is None or (right-left+1)<len(shortest):
                shortest=string[left:right+1]
            string_count[string[left]]-=1
            left+=1
    return shortest
m="ADOBECODEBANC" 
p="ABC"
print(shortest_substring(p,m))
```

```python
#house robbing
def houserob(arr):
    n=len(arr)
    if n==1:
        return arr[0]
    dp=[0]*n
    dp[0]=arr[0]
    dp[1]=max(arr[0],arr[1])
    for i in range(2,n):
        dp[i]=max(dp[i-1],arr[i]+dp[i-2])

    return dp[-1]
nums = [2, 7, 9, 3, 1]

print(houserob(nums))
```

```python
def coinChange(coins, amount):
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0

    for x in range(1, amount + 1):
        for coin in coins:
            if x >= coin:
                dp[x] = min(dp[x], dp[x - coin] + 1)

    return dp[amount] if dp[amount] != float('inf') else -1
```
