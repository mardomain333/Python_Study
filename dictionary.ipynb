Dictionary --


```python
dic={
    "gokul": 24,
    "siva": 33,
    "kumar": 12,                            
    "raja": 1,
    "akul": 20,
    "sara": 9,
    "ramu": 34   

}   
#to get all key values pairs as tuple
print(dic.items())

#use tuple unpacking for getting each key values
for key,value in dic.items():
    print(key,value)

#to get all keys
print(dic.keys())

#to get all values
print(dic.values())

#to add an item 
dic['key']='value'
print(dic)
#or
dic.update({'key':'value'})


#use pop to delete particular value using its key
print(dic.pop('key'))

dic.popitem() # pop the last key:value
print(dic)
#to get value using key
print(dic.get('kumar'))


###
###
# sorting a dictionary
sorted_dic=dict(sorted(dic.items())) #based on key
print(sorted_dic)
sorted_dic=dict(sorted(dic.items(),key=lambda x:x[1])) #based on values
print(sorted_dic)


```

```python
#swap key and values

swaped_dic={v:k for k,v in dic.items()}
print(swaped_dic)
```

Set

```python
set1={1,2,3,4,5}
set2={8,9,7}

set2.add(8) #add an item to set
print(set2)


diff=set1.difference(set2)  #set difference elements in set1 but not in set2
print(diff)

diff=set1.intersection(set2) #set intersection elements in both sets 
print(diff)

diff=set1.union(set2) #union the sets
print(diff)

#checking a set is subset of another
print("yes") if {0}.issubset(set1) else print("no")

if set1.issuperset({1}):
    print("yes") 

#removing an element(handle error)
set1.discard(2)
print(set1)

print("they are disjoint ") if set1.isdisjoint(set2) else print('not disjoint')



```

Some thing about * and **

```python
# * for list 

a,c,*b=1,2,3,4,5,5
print(a)
print(b)

def sumation(*args):
    print(args[0])
    s=sum(args)
    return s
print(sumation(1,2,3,0,4,5,))


def show(name,age,place):
    print(f"{name} of age {age} resides in {place}")

details=("gokul",3,"kollam")
a,b,c=details
print(a,b,c)
show(*details)
```

```python
# ** for dictionary

def show(**kwargs):
    print(kwargs)
    return (kwargs)

dic=show(name="gokul",age=23,place="kollam")

def show(name,age,place):
    print(name,age,place)
show(*dic) # 1* itta keys pass cheyum 
show(**dic)# 2* itta values pass cheyum
```

```python
def multi(*args):
    result=1
    for n in args:
        result*=n
    print(result)
lister=[2,3,4]
multi(*lister)
```
