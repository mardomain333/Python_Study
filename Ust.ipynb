Palindrome Score Calculator

Problem:
Given a string, increase score when palindromes of specific lengths are found.

+5 if a palindrome of length 4 exists

+10 if a palindrome of length 5 exists

Example:
Input → "ABABAAAA"
Output → 15
Explanation → "ABABA" gives 10, "AAAA" gives 5.


```python
def check_pali(str):
    if str==str[::-1]:
        return True
    else:
        return False

#checking for palidrome substring of lenght 4
s="ABABAAAA"
k=4
total=0
for i in range(len(s)-3):
    if check_pali(s[i:i+k]):
        total+=5
k=5
for i in range(len(s)-4):
    if check_pali(s[i:i+k]):
        total+=10
print(total)

```

```python
#two sum

target=5
l=[1,2,2,3,4,5]
coco=[5]
for a in l:
    doco=abs(target-a)
    if doco in l and doco not in coco:
        print(a,doco)
    coco.extend([a,doco])

```
